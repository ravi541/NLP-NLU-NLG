Now we will look into bi-gram,tri-gram and n-gram

```python
from nltk.util import bigrams, trigrams, ngrams
```

```python
string = 'generative ai and agentic ai is automate the industry & more focus on every domain'
quotes_tokens = nltk.word_tokenize(string)
quotes_tokens
```

```python
len(quotes_tokens)
```

```python
quotes_bigram=list(nltk.bigrams(quotes_tokens))
quotes_bigram
```

```python
quotes_trigram=list(nltk.trigrams(quotes_tokens))
quotes_trigram
```

```python
quotes_ngram=list(nltk.ngrams(quotes_tokens))
quotes_ngram
```

```python
quotes_ngram=list(nltk.ngrams(quotes_tokens,5))
quotes_ngram
```
